```python
import pandas as pd
import numpy as np
import seaborn as sns
sns.set()
import matplotlib.pyplot as plt

from sklearn import linear_model
import lightgbm as lgbm

from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.model_selection import KFold
```

```python
X_train = pd.read_csv('../data/X_train.csv',index_col='ROW_ID')
X_test = pd.read_csv('../data/X_test.csv',index_col='ROW_ID')

y_train = pd.read_csv('../data/y_train.csv',index_col='ROW_ID')
sample_submission = pd.read_csv('../data/sample_submission.csv',index_col='ROW_ID')
total_data = pd.concat([X_train,y_train],axis=1)
```

```python
ret_cols = [col for col in total_data.columns if 'RET' in col]
for i,col in enumerate(ret_cols):
    if i == 0 :
        continue
    else:
        momentum_col = f'MOMENTUM_{i}'
        X_train[momentum_col] = (X_train["RET_1"]/X_train[col])-1
```

```python
#corrilation of momentum features with target
momentum_cols = [col for col in X_train.columns if 'MOMENTUM' in col]
momentum_corr = X_train[momentum_cols].join(y_train).corr()['target'].sort_values(ascending=False)
print(momentum_corr)
```
